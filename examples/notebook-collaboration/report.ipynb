# Collaborative notebook

Synthetic data for live editing, selected-cell execution, and saved-output delivery.


```python
from IPython.display import display
import pandas as pd

values = pd.DataFrame({"group": ["A", "B", "C"], "value": [3, 7, 5]})
```

```python
display(values)
```

```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots(figsize=(6, 3))
ax.bar(values["group"], values["value"])
ax.set_title("Synthetic values")
display(fig)
plt.close(fig)
```

```python
from bokeh.io import output_notebook, show
from bokeh.plotting import figure
from bokeh.resources import INLINE

output_notebook(resources=INLINE, hide_banner=True)
plot = figure(x_range=values["group"].tolist(), height=300, width=600, title="Synthetic values", tools="pan,wheel_zoom,reset,save")
plot.vbar(x=values["group"].tolist(), top=values["value"].tolist(), width=0.6)
show(plot)
```

```python
import json

vega_spec = {
    "$schema": "https://vega.github.io/schema/vega-lite/v5.json",
    "description": "Synthetic values",
    "data": {"values": values.to_dict(orient="records")},
    "mark": "bar",
    "width": 500,
    "height": 280,
    "encoding": {"x": {"field": "group", "type": "nominal"}, "y": {"field": "value", "type": "quantitative"}},
}
# JupyterLab selects native Vega; Quarto uses the equivalent HTML representation.
vega_html = """<div id="collaboration-vega"></div>
<script>
// Quarto's notebook HTML host provides RequireJS.
requirejs.config({paths: {
  "vega": "https://cdn.jsdelivr.net/npm/vega@5.33.0/build/vega.min",
  "vega-lite": "https://cdn.jsdelivr.net/npm/vega-lite@5.23.0/build/vega-lite.min",
  "vega-embed": "https://cdn.jsdelivr.net/npm/vega-embed@6.29.0/build/vega-embed.min"
}});
requirejs(["vega-embed"], function(embed) {
  embed('#collaboration-vega', SPEC, {actions: false});
});
</script>""".replace("SPEC", json.dumps(vega_spec))
display({"application/vnd.vegalite.v5+json": vega_spec, "text/html": vega_html}, raw=True)
```
